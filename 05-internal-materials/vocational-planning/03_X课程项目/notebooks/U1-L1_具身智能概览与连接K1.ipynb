## 一、具身智能与人形机器人认知

本课先在“不驱动机器人运动”的前提下，建立对具身智能与人形机器人的整体认知，再进入 SDK 实操。


### 1.1 什么是具身智能（Embodied AI）

你可能已经对 **“人工智能”** 十分熟悉，比如与你对话的豆包，为你创作视频的Seedance，辅助学习\工作的Agent(WorkBuddy)等。它们是性能卓越的 **“数字大脑”**，能进行推理和生成。然而，它们都“藏在”屏幕后面的数字世界里，是一种 **“离身智能”**。

**具身智能**指“智能”不只存在于算法里，而是“长”在能感知、能行动的 **身体（body）** 上。通俗说：会看、会想、会动的机器人，才是具身智能，它们拥有物理身体，可以在真实的物理世界中进行感知、交互和学习。

人形机器人是其中最具代表性的载体——它用与人类相似的身体结构去理解世界、完成任务。


### 1.2 人形机器人的系统组成（四大子系统）

一台人形机器人可拆成四个相互协作的子系统，后面单元会逐一展开：

| 子系统 | 通俗称呼 | 组成与作用 |
|---|---|---|
| 感知 | “眼睛”和“前庭” | 相机、IMU、里程计、力传感器，负责采集外界与自身状态 |
| 计算 | “大脑” | 机载算力 + 算法（含大模型 / 控制算法），负责理解与决策 |
| 运动 | “肌肉和骨骼” | 关节电机、执行器，负责把决策变成实际动作 |
| 电源 | “心脏” | 电池与管理，决定续航与安全 |


### 1.3 具身智能的发展脉络（参考）

| 阶段 | 约略时期 | 代表进展 | 对本课学习者的意义 |
|---|---|---|---|
| 符号智能 | 1950s–1980s | 智能被理解为“符号推理”，与身体无关 | 理解“离身智能”的局限，反衬具身路线的价值 |
| 行为主义与机器人学 | 1980s–2000s | 强调“感知—行动”闭环，真实机器人进入研究 | 机器人开始摆脱纯仿真，走向真实物理交互 |
| 深度学习赋能感知 | 2010s | 视觉 / 语音识别突破，机器人“看得清、听得懂” | 感知能力快速普及，为具身智能打底 |
| 大模型与 VLA | 2020s 至今 | 大模型带来通用推理，VLA 把“看—想—动”连成一体 | 开发者可用统一接口调度“感知—决策—执行” |
### 1.3.1 参考文献（延伸阅读）

本脉络涉及的代表性文献，供学有余力者延伸了解，**不纳入考核**：
- Newell & Simon (1976). *Computer Science as Empirical Inquiry: Symbols and Search*. 将智能理解为“符号推理”，是“离身智能”路线的代表。
- Brooks (1991). *Intelligence without Representation*. 提出“无表征”的行为主义思路，强调“感知—行动”闭环。
- Krizhevsky et al. (2012). *ImageNet Classification with Deep Convolutional Neural Networks*（AlexNet）. 深度学习赋能感知的里程碑。
- Driess et al. (2023). *PaLM-E: An Embodied Multimodal Language Model*. 将多模态感知直接接入语言模型，迈向“看—想—动”一体。
- Brohan et al. (2023). *RT-2: Vision-Language-Action Models Transfer Web Knowledge to Robotic Control*. VLA 端到端把“看+说”变成“动”的代表工作。


### 1.4 什么是 SDK 与 BoosterRobot

SDK（Software Development Kit，软件开发工具包）是 Booster Robot 公司提供的“遥控器说明书 + 万能钥匙”，把操作机器人的能力封装成一行行可调用的接口；本课使用的 SDK 名为 **BoosterOS**（Python 包 `boosteros`）。

`BoosterRobot` 是操作 K1 的统一入口：一行代码即可建立连接，之后所有读 / 写机器人的能力都通过它调用。本课以“只读”方式使用它，已经能体会到“用程序操控实体机器人”的基本范式。

下面用 `BoosterRobot` 连接 K1，并逐项读取它的“身份证”与“身体清单”。


### 2.1 软件环境与连接准备（BoosterStudio）

本环节在动手写代码前，需要先让电脑与 K1 建立通信。用到的软件与操作包括：

- **BoosterStudio**：适配 K1 的具身智能应用开发软件，主要功能有——查看机器人连接状态、实时显示关节 / 姿态、手动切换运行模式、发送简单控制指令。它是确认“机器人已就绪、可连接”的常用入口。
- **配网与查 IP**：将 K1 与电脑接入同一 WiFi，并查到 K1 的 IP 地址，配网需要使用 Booster APP，查询 IP 地址可以使用 APP 或者 Studio, 只有完成配网我们才可以通过 IP 连接 K1 机器人并完成相应程序开发。
- **安装 SDK**：在电脑终端执行 `pip install boosteros`，获得 `BoosterRobot` 等接口。

> 📎 **附件提示**：本环节涉及的 BoosterStudio 软件界面、K1 实物示意图与逐步连接的操作步骤，详见附件《K1 连接与 BoosterStudio 操作图示》，请结合图文完成准备。


### 2.2 步骤 1：连接 K1 机器人

`BoosterRobot` 是操作 K1 的统一入口。运行下面单元格建立连接。

- 若连接失败，请检查：K1 是否开机、电脑与 K1 是否同一 WiFi、SDK 是否已安装。
- 若提示 `ModuleNotFoundError: No module named 'boosteros'`，代表 boosteros 尚未安装。

```python
from boosteros.robots.booster import BoosterRobot

robot = BoosterRobot()
print("已连接 K1，开始探索 🤖")
```

### 2.3 步骤 2：读取机器人身份信息

`robot.robot_info` 返回一个只读的元信息对象，包含制造商、型号、序列号、固件版本等。


```python
info = robot.robot_info
print(f"机器人的元信息：{info}")
```

`RobotInfo` 表示机器人的基础信息，具体包括：

| 属性　　　　　　 | 类型　　　　　　　| 说明　　　　　　　　　　　　　 |
| ------------------| -------------------| --------------------------------|
| manufacturer　　 | str　　　　　　　 | 机器人厂家名称　　　　　　　　 |
| model　　　　　　| str　　　　　　　 | 机器人型号；不可用时为空字符串 |
| name　　　　　　 | str　　　　　　　 | 机器人名称；不可用时为空字符串 |
| serial_number　　| str　　　　　　　 | 序列号；不可用时为空字符串　　 |
| firmware_version | str　　　　　　　 | 固件版本号；不可用时为空字符串 |
| extra　　　　　　| dict[str, object] | 扩展信息　　　　　　　　　　　 |

**其中比较有价值的字段：**：
- **型号（model）**：对应机器人的硬件规格与能力边界，不同型号对应的传感器、执行器都不同。
- **序列号（serial_number）**：机器人的唯一身份标识（“身份证号”），多机并存时可据此区分、定位具体设备，使用BoosterStudio连接机器人时也可以根据序列号寻找自己的机器人。
- **固件版本（firmware_version）**：底层控制软件的版本号，不同版本下接口行为、动作库、参数范围可能存在差异；运行前确认版本有助于统一实验环境、稳定复现结果。

```python
print("型号：    ", info.model)
print("序列号：  ", info.serial_number)
print("固件版本：", info.firmware_version)
```

### 2.4 步骤 3：浏览关节清单

`robot.list_joints()` 返回所有关节的信息，可以打印查看。


```python
joints = robot.list_joints()
print(f"关节总数：{len(joints)}")
print(f"list_joints{joints}")
```

K1 的 `list_joints()` 包含 22 个关节，每个关节包括：

| 属性 | 类型 | 说明 |
|---|---|---|
| name | str | 关节名称 |
| limits | JointLimits | 关节位置范围（min / max，单位 rad） |
| max_torque | float | 关节最大力矩 |
| max_velocity | float | 关节最大速度（单位 rad/s） |
| extra | dict[str, object] | 扩展信息 |

观察关节的**数量与命名规律**（头部、手臂、腿部等），即可直观理解 K1 的“身体结构”。


```python
for j in joints:
    print(f"{j.name:<20} 限位: [{j.limits.min:.3f}, {j.limits.max:.3f}] rad  最大速度: {j.max_velocity:.3f} rad/s")
```

### 2.5 步骤 4：浏览预定义动作库

`robot.list_actions()` 返回出厂预置的动作列表（如挥手、鞠躬、踢球等）。


```python
actions = robot.list_actions()
print(f"预定义动作总数：{len(actions)}")
print(f"预定义动作内容：{actions}")
```

K1 包含多个预定义动作，每个动作包括：

| 属性 | 类型 | 说明 |
|---|---|---|
| id | str | 动作 ID，可作为 `robot.do_action(action_id)` 的 action_id 参数 |
| type | str | 动作类型，如 "upper_body"、"whole_body" |
| duration | float | 预估持续时间，`-1` 表示不确定 |
| interruptible | bool | 是否可被新指令打断 |


```python
for a in actions:
    dur = f"{a.duration:.1f}s" if a.duration is not None else "-"
    print(f"动作名称：{a.id:<25} 类型：{a.type:<15} 时长：{dur:<8} 可中断：{'是' if a.interruptible else '否'}")
```

### 2.6 步骤 5：查看当前状态

机器人开机后一般拥有多个不同的运行状态。
- `get_mode()` 返回当前运行模式；
- `get_battery()` 返回电量信息。

```python
print("当前模式：", robot.get_mode())
print("电池信息：", robot.get_battery())
```
