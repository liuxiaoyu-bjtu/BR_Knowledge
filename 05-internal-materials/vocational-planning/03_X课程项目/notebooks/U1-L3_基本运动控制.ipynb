## 步骤 2：运动坐标系与速度指令

K1 使用**机器人自身坐标系**：

- `x` 轴：机器人正前方
- `y` 轴：机器人左侧
- `z` 轴：竖直向上（右手系）

速度指令 `set_velocity(vx, vy, vyaw)` 的三个参数均相对机器人自身坐标系：

- `vx`：前后速度（m/s），**正值 = 前进**
- `vy`：左右平移速度（m/s），**正值 = 向左**
- `vyaw`：转向角速度（rad/s），**正值 = 逆时针（向左转）**

这些速度描述的是“机器人想往哪个方向走”，不是相对地面的绝对方向。
