# U8-L4：成果展演与答辩

**课程目标**

1. 能对照本组需求演示导览或踢球作品，用程序与现场证据说明完成情况。
2. 能结合实际代码或彩排现象解释两个本组用到的概念，并说明模块交接和修改依据。
3. 能根据评价指出作品的使用条件、未解决项及下一步可验证的改进。

## 一、任务：把作品交给能按说明使用的人

前三课已经留下需求、程序、测试与复测。本课把这些材料组织成一次演示：说明作品服务谁，现场运行已经验证的版本，用记录解释完成情况和限制，再回答实现问题。

### 1.1 一组八分钟怎样安排

每组五分钟演示、三分钟答辩。五分钟内依次用约一分钟交代场景和条件，原则上两分钟内运行机器人，其余时间展示结果、修改证据并复位。不要为了凑满时间重复动作。

### 1.2 全班怎样轮换

45分钟按四组安排：核查5分钟、四组各8分钟共32分钟、评价5分钟、收尾3分钟。组数多时由教师分场；有足够设备与指导人员时可分站。候场组检查材料，不抢连演示中的机器人；一台机器人每轮只由指定操作员控制。

**任务01：先按文件检查，再按场地检查。** 操作员打开L3本组目录，加载已定稿版本并核对版本号；本课不提供另一套运动代码，仍使用L3的STEP 01～03和STEP 06。只运行已验证用例，不点击“全部运行”。加载连接前先确认旧Notebook资源已退出。

| 主题、需求和验收表 | | |
| 本组程序、接口和分工 | | |

若当场缺少必要条件，向教师说明，保留未运行项；历史视频可辅助解释，但不能标成这次现场运行结果。

## 三、按需求演示，再用证据答辩

### 3.1 准备并执行五分钟演示

**任务02：按以下顺序排一次讲解。**

1. 讲解人说明对象、场地和一条核心需求，指出运行前条件。
2. 用本组流程图指出观察、介绍、新观测和行为怎样交接，以及本组具体改了哪里。
3. 操作员在本次条件获确认后，沿用L3 STEP 03运行已验证用例。记录员把本次程序结果与现场现象分别填表，保护同学全程就位。
4. 运行结束先按L3 STEP 06核查退出，再人工复位；若未完成则按既定流程结束，说明已取得的记录，不为挽回演示连续启动。

## 四、评价与反馈

**任务03：根据同一组证据评分。** 满分100，按以下四项记录。动作复杂、幅度大或使用更多API不单独加分。

## 五、个人说明与最终交付

**任务04：每人填写，并指向本组证据。**

### 思考题与个人说明

1. 我负责哪条需求、哪个函数或哪份记录？交给谁使用？
2. 我能解释哪处代码或哪次修改？在哪个版本、哪个用例中看到结果？
3. 作品适用的场地、姿态与材料是什么？离开这些条件后先补哪项验证？

展演后核对线程、音频、动作与管理器退出，保留未结束资源的原对象并交由教师处理。先准备模式等站稳，再按教师安排人工支撑坐下或关机。

按下面清单整理最终文件，另一位成员照运行说明核对一遍：

```text
本组交付/
├── u8_team.py、x_lab/、标定文件（如用）
├── L3运行Notebook与运行说明
├── 需求与接口、分工、已知问题
├── 模块测试、基线、复测JSON及现场表
└── 演示记录、视频（如有）、评价和个人说明
```

不把仍待测试的功能写成已具备能力。视频补充现场证据，替代现场演示须按教师安排，来源和时间应注明。

**完成核对**：文件和版本可交接，需求有对应证据，未完成项明确，现场资源全部核查，使用者能按说明找到启动、停止和复位步骤。

### 检查点

- [__] 能按本组需求演示作品，用程序与现场记录说明完成程度。
- [__] 能结合本组代码或现象解释两个概念及个人职责、模块交接和修改依据。
- [__] 能整理交付版本与运行说明，列明适用条件、未解决项和待补用例。
