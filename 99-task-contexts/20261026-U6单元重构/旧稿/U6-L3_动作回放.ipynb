## 二、回放＝重演轨迹

回放一条轨迹，就是让机器人把录制时的关节角度序列再走一遍。录制时每个关节在什么时刻转到什么角度，回放时照着再来一遍，动作就「重演」出来了。

回放前机器人要处于**行走模式**——回放要驱动关节运动，和行走模式下的运动控制是同一套。

## 三、回放自己录的轨迹

⚠️ **安全**：回放会让机器人动起来。请确认机器人周围空旷、留人在旁；回放的动作如果幅度大，人站到它够不到的位置。出现意外直接通过按钮切到**准备模式**。

📌 **前置**：上节课保存的轨迹文件（`my_action.btraj`）还在当前工作目录；或准备好教师提供的轨迹文件。

**STEP 01：** 连接机器人

```python
from boosteros.robots.booster import BoosterRobot
from boosteros.types import TrajectoryData

robot = BoosterRobot()

print("已连接：", robot.robot_info.model, robot.robot_info.serial_number)
print("当前模式：", robot.get_mode())
```

**开始下面程序前，机器人应处于行走模式。** 如果机器人当前不在行走模式：先用按钮把机器人切到准备模式，等它站直站稳；再切到行走模式。确认进入行走模式后，再运行下面的代码。

**STEP 02：** 读入轨迹并回放

<font color="red">提示：</font>把要回放的轨迹文件路径填到下面代码的引号里（上节课保存的是 `my_action.btraj`）。回放开始后不要靠近机器人，等它自己停下来。

```python
traj = TrajectoryData.load("my_action.btraj")   # 把要回放的轨迹路径填到这里
print("轨迹关节名：", traj.joint_names)
print("轨迹时长：", traj.duration.seconds, "秒")

handle = robot.execute_trajectory(traj)
print("开始回放...")
status = handle.wait()
print("回放结束，状态：", status)
```

## 四、预定义动作

### 4.1 预定义动作是厂商固化的

回放的轨迹是**你自己录的**。还有一类动作，是厂商已经做好、固化在机器人里的，直接调用就行，比如挥手、跳舞、踢球。在 BoosterOS SDK 里，这类动作叫**预定义动作**，接口是 `do_action()`。

`do_action()` 在 U1-L4《头部、步态与预定义动作》里已经用过：先用 `list_actions()` 看清单，再挑一个动作 id 传给 `do_action()`。本课把它用起来，多看几种动作的差别。

清单里每一项有几个字段：`id`（动作编号，传给 `do_action()`）、`type`（`upper_body` 只动上身 / `whole_body` 全身动）、`duration`（时长，`-1` 表示不确定）、`interruptible`（能不能被打断）。

**STEP 03：** 读动作清单

```python
actions = robot.list_actions()
print("一共有", len(actions), "个预定义动作")

for a in actions:
    print(f"{a.id:<24} 类型 {a.type:<12} 时长 {a.duration} 可打断 {a.interruptible}")
```

**STEP 04：** 挑 3 个预定义动作分别调用

<font color="red">提示：</font>从上面清单里挑 3 个动作 id，替换下面三个下划线占位符，再运行。每执行一个，观察它动了哪些部位、大概多长。

```python
action_ids = ["____", "____", "____"]   # 从清单里挑 3 个动作 id 填进来

for action_id in action_ids:
    print("执行：", action_id)
    handle = robot.do_action(action_id)
    handle.wait()
    print("结束：", action_id)
```

## 五、收尾

实验做完，机器人还停在行走模式。收尾顺序：先用按钮把机器人切回**准备模式**，等它站直稳住；再由人工扶着切到**阻尼模式**（damping），扶机器人坐下，确认坐稳。不要从行走模式直接切到阻尼模式（那样机器人会当场瘫倒）。
